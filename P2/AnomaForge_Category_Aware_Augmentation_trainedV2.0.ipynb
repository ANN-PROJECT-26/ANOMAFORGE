# AnomaForge — Category-aware augmentation

This notebook trains one student on normal images from all 15 MVTec AD categories. Stage 4B uses an editable, category-specific augmentation policy and passes the same randomly transformed image to the teacher and student. Development images, threshold calibration, and test images are not augmented by this training step.

The policies are conservative starting points. Check that each selected transformation is normal for your own inspection setup; rotation can introduce white corners. Preview the categories before training.

The category-aware run has its own experiment ID and checkpoint files. A matching run can resume automatically. To continue from a previous compatible model with a different augmentation policy, select its checkpoint in Stage 6; this starts a new run from its trained weights with a fresh optimizer and history. The notebook will not start from scratch unless you deliberately change the code. Old Drive checkpoints remain untouched.

## Stage 4 — Split normal training images across all categories

Each category is divided into **80% training and 20% development images** using seed 42. All category folders must be present for the shared all-category student. No official MVTec test image is included in either partition.

```python
import random
import torch
import torch.nn.functional as F
from PIL import Image

SPLIT_SEED = 42
missing_categories = sorted(set(ALL_CATEGORIES) - set(available_categories))
if missing_categories:
    raise FileNotFoundError(
        'All-category training requires all 15 MVTec AD categories. '
        'Missing: ' + ', '.join(missing_categories)
        + '. Place each category beside bottle/ under the MVTec root.'
    )

category_train_paths, category_dev_paths = {}, {}
train_paths, dev_paths = [], []

for category in ALL_CATEGORIES:
    originals = image_files(MVTEC_ROOT / category / 'train' / 'good')
    if len(originals) < 2:
        raise ValueError(f'{category}: at least two normal images are needed.')
    shuffled = originals.copy()
    random.Random(SPLIT_SEED).shuffle(shuffled)
    dev_count = max(1, round(len(originals) * 0.20))
    category_train_paths[category] = shuffled[dev_count:]
    category_dev_paths[category] = shuffled[:dev_count]
    train_paths.extend(category_train_paths[category])
    dev_paths.extend(category_dev_paths[category])
    print(f'{category:13s}: {len(category_train_paths[category]):3d} train | '
          f'{len(category_dev_paths[category]):3d} dev')

assert set(train_paths).isdisjoint(dev_paths)
print(f'\nAll categories: {len(train_paths)} training / {len(dev_paths)} development images')
print('Official test images used in training: 0')

teacher.model.eval()
assert all(not p.requires_grad for p in teacher.model.parameters())
print('Teacher frozen:', True)
```

## Stage 4B — Category-aware training augmentation



```python
%%writefile /content/trn_img_augmentation_category.py
"""Category-aware, on-the-fly augmentation for MVTec normal training images.

The policies below are conservative starting points, not validated guarantees.
Review each category and keep only transformations that are normal for your
inspection setup. Development and test images should not be augmented here.
"""

import random
from pathlib import Path

from PIL import Image, ImageEnhance, ImageOps


def policy(flip=0.0, rotation=0.0, degrees=0.0, brightness=0.2,
           brightness_min=0.97, brightness_max=1.03):
    """Build one editable set of probabilities and transformation limits."""
    return {
        'horizontal_flip_probability': flip,
        'rotation_probability': rotation,
        'max_rotation_degrees': degrees,
        'brightness_probability': brightness,
        'brightness_min': brightness_min,
        'brightness_max': brightness_max,
    }


AUGMENT_CONFIG = {
    'enabled': True,
    'seed': 42,
    'categories': {
                   # Variable-pose objects
                   'bottle':     policy(flip=0.5, rotation=0.35, degrees=7),
                   'hazelnut':   policy(flip=0.35, rotation=0.35, degrees=8),
                   'metal_nut':  policy(rotation=0.35, degrees=8),
                   'screw':      policy(rotation=0.25, degrees=6),

                   # Textures
                   'carpet':     policy(flip=0.35),
                   'leather':    policy(flip=0.35),
                   'tile':       policy(flip=0.35),
                   'grid':       policy(),
                   'wood':       policy(),

                   # Start with brightness changes only
                   'cable':      policy(),
                   'capsule':    policy(),
                   'pill':       policy(),
                   'toothbrush': policy(),
                   'transistor': policy(),
                   'zipper':     policy(),
                },
}

# Kept for interactive use; the notebook supplies its own per-epoch RNG.
augmentation_rng = random.Random(AUGMENT_CONFIG['seed'])


def category_from_path(image_path):
    """Read the category from .../<category>/train/good/<image>."""
    path = Path(image_path)
    if path.parent.name != 'good' or path.parent.parent.name != 'train':
        raise ValueError(f'Expected <category>/train/good/<image>: {path}')
    category = path.parent.parent.parent.name
    if category not in AUGMENT_CONFIG['categories']:
        raise ValueError(f'No augmentation policy for category: {category}')
    return category


def augment_normal_image(image, rng, category=None):
    """Return an RGB view using this category's independently sampled transforms.

    category must be passed when augmentation is enabled; this prevents older
    notebook callers from silently using the wrong category policy.
    """
    result = image.convert('RGB')
    if not AUGMENT_CONFIG['enabled']:
        return result
    if category not in AUGMENT_CONFIG['categories']:
        raise ValueError(
            f'Unknown or missing augmentation category: {category!r}. '
            'Pass category=category_from_path(image_path).'
        )

    cfg = AUGMENT_CONFIG['categories'][category]
    if rng.random() < cfg['horizontal_flip_probability']:
        result = ImageOps.mirror(result)

    if rng.random() < cfg['rotation_probability']:
        angle = rng.uniform(-cfg['max_rotation_degrees'], cfg['max_rotation_degrees'])
        result = result.rotate(
            angle,
            resample=Image.Resampling.BILINEAR,
            expand=False,
            fillcolor=(255, 255, 255),
        )

    if rng.random() < cfg['brightness_probability']:
        factor = rng.uniform(cfg['brightness_min'], cfg['brightness_max'])
        result = ImageEnhance.Brightness(result).enhance(factor)

    return result


def preview_augmentation(image_path, category=None):
    """Show random views of one normal image without advancing training RNG."""
    import matplotlib.pyplot as plt

    category = category or category_from_path(image_path)
    if category not in AUGMENT_CONFIG['categories']:
        raise ValueError(f'No augmentation policy for category: {category}')
    with Image.open(image_path) as image:
        original = image.convert('RGB')

    preview_rng = random.Random(123)
    examples = [('Original', original)]
    for number in range(1, 8):
        view = augment_normal_image(original, preview_rng, category=category)
        examples.append((f'Random view {number}', view))

    fig, axes = plt.subplots(2, 4, figsize=(14, 7))
    for axis, (title, image) in zip(axes.flat, examples):
        axis.imshow(image)
        axis.set_title(title)
        axis.axis('off')
    fig.suptitle(f'{category}: augmentation preview')
    plt.tight_layout()
    plt.show()
    print('Category:', category, '| Training example:', Path(image_path).name)
    print('Settings:', AUGMENT_CONFIG['categories'][category])
```

```python
import importlib.util
from pathlib import Path

# The preceding cell writes the same category-aware module to this runtime.
AUGMENTATION_FILE = Path('/content/trn_img_augmentation_category.py')
if not AUGMENTATION_FILE.is_file():
    raise FileNotFoundError('Run the category augmentation module cell first.')

spec = importlib.util.spec_from_file_location('anomaforge_p2_augmentation', AUGMENTATION_FILE)
if spec is None or spec.loader is None:
    raise ImportError(f'Unable to load {AUGMENTATION_FILE}')
augmentation_module = importlib.util.module_from_spec(spec)
spec.loader.exec_module(augmentation_module)

AUGMENT_CONFIG = augmentation_module.AUGMENT_CONFIG
augment_normal_image = augmentation_module.augment_normal_image
category_from_path = augmentation_module.category_from_path
preview_augmentation = augmentation_module.preview_augmentation

# Change to False to use the unaugmented baseline configuration.
AUGMENT_CONFIG['enabled'] = True
SHOW_AUGMENTATION_PREVIEW = True
PREVIEW_CATEGORY = 'bottle'  # Any available training category.

print('Augmentation module:', AUGMENTATION_FILE)
print('Category-aware training augmentation enabled:', AUGMENT_CONFIG['enabled'])
print('Number of category policies:', len(AUGMENT_CONFIG['categories']))
if SHOW_AUGMENTATION_PREVIEW:
    preview_augmentation(category_train_paths[PREVIEW_CATEGORY][0])
```

## Stage 5 — Verify that student gradients work

This checks one backward pass on a **temporary** student. The main student is not updated here, so the full training starts from clean initial weights or a saved checkpoint.

```python
# Use a disposable student for this one-step gradient sanity check.
torch.manual_seed(SPLIT_SEED)
probe_student = StudentCNN().to(teacher.device)
probe_optimizer = torch.optim.Adam(probe_student.parameters(), lr=1e-3)
probe_student.train()

# Use an independent RNG so this check cannot change the training sequence.
with Image.open(train_paths[0]) as image:
    probe_image = image.convert('RGB')
if AUGMENT_CONFIG['enabled']:
    probe_rng = random.Random(int(AUGMENT_CONFIG['seed']) + 10_000)
    probe_category = category_from_path(train_paths[0])
    probe_image = augment_normal_image(probe_image, probe_rng, category=probe_category)
teacher_pixels, teacher_features = teacher.extract(probe_image)

probe_optimizer.zero_grad(set_to_none=True)
probe_features = probe_student(teacher_pixels)
assert probe_features.shape == teacher_features.shape

probe_loss = F.mse_loss(
    F.normalize(probe_features, p=2, dim=1),
    F.normalize(teacher_features, p=2, dim=1)
)
first_parameter = next(probe_student.parameters())
weights_before = first_parameter.detach().clone()
probe_loss.backward()
gradient_norm = first_parameter.grad.norm().item()
probe_optimizer.step()
weight_change = (first_parameter.detach() - weights_before).abs().max().item()

print('Teacher / student features:', tuple(teacher_features.shape))
print('One-step loss:', probe_loss.item())
print('Gradient norm:', gradient_norm)
print('Largest sample weight update:', weight_change)
assert gradient_norm > 0 and weight_change > 0

del probe_student, probe_optimizer, probe_features, probe_loss
```

## Stage 6 — Resume training or initialize from a previous model

With a matching category-aware checkpoint, the cell below restores its model, optimizer, and epoch. When switching from an older policy, copy the full path of your chosen `_last.pt` checkpoint into `WARM_START_CHECKPOINT`. The notebook verifies its dataset, teacher, and student configuration and starts a separate category-aware run from those trained weights; its optimizer, epoch count, and loss history start fresh.

Leave the same source checkpoint selected when returning to this category-aware run so that its ID remains stable. If no matching run or valid source checkpoint is found, training is stopped. Previous checkpoints are not overwritten.


```python
# Saved student weights and training-state management (Google Drive).
import copy
import hashlib
import inspect
import json
import importlib.util
from pathlib import Path

# False until a compatible checkpoint has actually been loaded.
training_initialized = False
BATCH_SIZE = 4
EPOCHS = 10  # Target epochs for this run, not additional epochs after resume.
LEARNING_RATE = 1e-3
FORCE_RETRAIN = False
REQUIRE_GPU_FOR_TRAINING = True

# Set to the full path of a trusted earlier *_last.pt to start the NEW
# category-aware experiment from its trained weights. Keep the same source
# selected when returning to this run. Leave None to resume a run created
# without a warm-start source (or to find out which checkpoints are present).
WARM_START_CHECKPOINT = Path('/content/drive/MyDrive/ANOMAFORGE/mvtec_anomaly_detection/checkpoints/student_all_15_cat_aug_44d5b6872a50_last.pt')

def relative_image_id(path):
    return Path(path).relative_to(MVTEC_ROOT).as_posix()

def file_sha256(path):
    digest = hashlib.sha256()
    with Path(path).open('rb') as source:
        for chunk in iter(lambda: source.read(1024 * 1024), b''):
            digest.update(chunk)
    return digest.hexdigest()

run_config = {
    'categories': list(ALL_CATEGORIES),
    'train_ids': sorted(relative_image_id(p) for p in train_paths),
    'dev_ids': sorted(relative_image_id(p) for p in dev_paths),
    'split_seed': SPLIT_SEED,
    'teacher_model': 'facebook/dinov3-vits16-pretrain-lvd1689m',
    'teacher_code_sha256': file_sha256(inspect.getfile(DINOv3Teacher)),
    'student_code_sha256': file_sha256(inspect.getfile(StudentCNN)),
    'input_size': 224,
    'loss': 'MSE of L2-normalized patch feature maps',
    'batch_size': BATCH_SIZE,
    'learning_rate': LEARNING_RATE,
    'checkpoint_schema': 1,
}
augmentation_enabled = bool(AUGMENT_CONFIG['enabled'])
if augmentation_enabled:
    # Deep copy freezes nested category settings for later change detection.
    run_config['training_augmentation'] = {
        'settings': copy.deepcopy(AUGMENT_CONFIG),
        'module_sha256': file_sha256(AUGMENTATION_FILE),
        'rng_scheme': 'python_random_seed_plus_epoch_v1',
    }

warm_start_path = None
if WARM_START_CHECKPOINT:
    warm_start_path = Path(WARM_START_CHECKPOINT).expanduser()
    if not warm_start_path.is_file():
        raise FileNotFoundError('Warm-start checkpoint not found: ' + str(warm_start_path))
    # Different initial weights must not share an experiment ID.
    run_config['warm_start_checkpoint_sha256'] = file_sha256(warm_start_path)

run_id = hashlib.sha256(json.dumps(run_config, sort_keys=True).encode()).hexdigest()[:12]
checkpoint_dir = PROJECT_DIR / 'checkpoints'
checkpoint_dir.mkdir(parents=True, exist_ok=True)
checkpoint_prefix = 'student_all_15_cat_aug' if augmentation_enabled else 'student_all_15'
best_path = checkpoint_dir / f'{checkpoint_prefix}_{run_id}_best.pt'
last_path = checkpoint_dir / f'{checkpoint_prefix}_{run_id}_last.pt'
print('Category-aware augmentation:', 'enabled' if augmentation_enabled else 'disabled')
print('Experiment ID:', run_id)
print('Best checkpoint:', best_path)
print('Last checkpoint:', last_path)

torch.manual_seed(SPLIT_SEED)
student_model = StudentCNN().to(teacher.device)
optimizer = torch.optim.Adam(student_model.parameters(), lr=LEARNING_RATE)
train_history, dev_history = [], []
completed_epochs, best_epoch = 0, 0
best_dev_loss = float('inf')

if FORCE_RETRAIN:
    raise RuntimeError('FORCE_RETRAIN is disabled here to protect existing work.')

restore_from = last_path if last_path.is_file() else best_path
if restore_from.is_file():
    state = torch.load(restore_from, map_location=teacher.device, weights_only=True)
    if state.get('run_config') != run_config:
        raise RuntimeError('Stored checkpoint does not match this experiment.')
    if state.get('optimizer_state_dict') is None:
        raise RuntimeError('This checkpoint has no optimizer state for an exact resume.')
    student_model.load_state_dict(state['model_state_dict'], strict=True)
    optimizer.load_state_dict(state['optimizer_state_dict'])
    completed_epochs = int(state['epoch'])
    best_epoch = int(state['best_epoch'])
    best_dev_loss = float(state['best_dev_loss'])
    train_history = list(state['train_history'])
    dev_history = list(state['dev_history'])
    training_initialized = True
    print(f'Restored {restore_from.name} at epoch {completed_epochs}.')
elif warm_start_path is not None:
    # A changed augmentation policy is a new experiment, not an exact resume.
    old = torch.load(warm_start_path, map_location=teacher.device, weights_only=True)
    old_config = old.get('run_config')
    if not isinstance(old_config, dict) or 'model_state_dict' not in old:
        raise RuntimeError('Source is not a compatible full notebook checkpoint.')
    required_keys = (
        'categories', 'train_ids', 'dev_ids', 'split_seed', 'teacher_model',
        'teacher_code_sha256', 'student_code_sha256', 'input_size', 'loss',
    )
    mismatch = [key for key in required_keys if old_config.get(key) != run_config.get(key)]
    if mismatch:
        raise RuntimeError('Warm-start checkpoint differs in: ' + ', '.join(mismatch))
    student_model.load_state_dict(old['model_state_dict'], strict=True)
    training_initialized = True
    print(f'Loaded student weights from {warm_start_path.name} '
          f'(source epoch {old.get("epoch", "unknown")}).')
    print('New category-aware run: epoch count, optimizer, and history start fresh.')
else:
    candidates = sorted(p.name for p in checkpoint_dir.glob('*.pt'))
    print('Available checkpoint files:', candidates or '(none found)')
    raise FileNotFoundError(
        'No matching category-aware checkpoint. Set WARM_START_CHECKPOINT to '
        'the full path of a compatible earlier _last.pt, then rerun this cell. '
        'Training has not started.'
    )

print(f'Epochs completed in this run: {completed_epochs}; target: {EPOCHS}')
if completed_epochs >= EPOCHS:
    print('Training is already complete; the next cell will skip it.')
```

```python
from torch.utils.data import DataLoader

if not globals().get('training_initialized', False):
    raise RuntimeError('No compatible checkpoint was loaded. Run Stage 6 setup first.')


def load_images(paths, augmentation_rng=None):
    """Load RGB images; apply augmentation only when an RNG is supplied."""
    images = []
    for path in paths:
        with Image.open(path) as image:
            rgb = image.convert('RGB')
        if augmentation_rng is not None:
            category = category_from_path(path)
            rgb = augment_normal_image(rgb, augmentation_rng, category=category)
        images.append(rgb)
    return images


def feature_matching_loss(student_features, teacher_features):
    if student_features.shape != teacher_features.shape:
        raise ValueError('Student/teacher feature shapes differ.')
    return F.mse_loss(
        F.normalize(student_features, p=2, dim=1),
        F.normalize(teacher_features, p=2, dim=1)
    )


def run_epoch(paths, training, epoch):
    # A changed configuration needs a fresh experiment ID/checkpoint setup.
    if bool(AUGMENT_CONFIG['enabled']) != augmentation_enabled or (
        augmentation_enabled and dict(AUGMENT_CONFIG) != run_config['training_augmentation']['settings']
    ):
        raise RuntimeError('Augmentation settings changed. Rerun the Stage 6 checkpoint setup cell.')
    if augmentation_enabled and file_sha256(AUGMENTATION_FILE) != run_config['training_augmentation']['module_sha256']:
        raise RuntimeError('Augmentation module changed. Rerun the module and Stage 6 setup cells.')

    student_model.train(training)
    # Per-epoch seeds make a resumed run repeat the same epoch's random views.
    # Preview and one-step checks use independent generators.
    augmentation_rng = (
        random.Random(int(AUGMENT_CONFIG['seed']) + epoch)
        if training and augmentation_enabled else None
    )
    loader = DataLoader(
        [str(path) for path in paths],
        batch_size=BATCH_SIZE,
        shuffle=training,
        generator=torch.Generator().manual_seed(SPLIT_SEED + epoch),
        num_workers=0,
    )
    loss_sum, image_count = 0.0, 0
    for batch_paths in loader:
        # Development passes augmentation_rng=None and stays unaugmented.
        images = load_images(batch_paths, augmentation_rng=augmentation_rng)
        # The teacher sees the transformed images once and returns the exact
        # preprocessed pixels also used by the student.
        pixels, teacher_features = teacher.extract(images)  # Frozen teacher
        if training:
            optimizer.zero_grad(set_to_none=True)
        with torch.set_grad_enabled(training):
            student_features = student_model(pixels)
            loss = feature_matching_loss(student_features, teacher_features)
            if not torch.isfinite(loss).item():
                raise RuntimeError('Non-finite feature-matching loss.')
            if training:
                loss.backward()
                optimizer.step()
        loss_sum += loss.item() * len(images)
        image_count += len(images)
    return loss_sum / image_count


def save_training_state(path, epoch):
    """Save weights and optimizer/history after each epoch on persistent Drive."""
    state = {
        'model_state_dict': student_model.state_dict(),
        'optimizer_state_dict': optimizer.state_dict(),
        'epoch': epoch,
        'best_epoch': best_epoch,
        'best_dev_loss': best_dev_loss,
        'train_history': train_history,
        'dev_history': dev_history,
        'run_config': run_config,
    }
    temporary = path.with_name(path.name + '.tmp')
    torch.save(state, temporary)
    temporary.replace(path)


if completed_epochs < EPOCHS and REQUIRE_GPU_FOR_TRAINING and teacher.device.type != 'cuda':
    raise RuntimeError('Select a GPU runtime in Colab before training all 15 categories.')

if completed_epochs >= EPOCHS:
    print(f'Skipping student training: {completed_epochs} epochs already saved.')
else:
    for epoch in range(completed_epochs + 1, EPOCHS + 1):
        train_loss = run_epoch(train_paths, training=True, epoch=epoch)
        dev_loss = run_epoch(dev_paths, training=False, epoch=epoch)
        train_history.append(train_loss)
        dev_history.append(dev_loss)
        print(f'Epoch {epoch:02d}/{EPOCHS}: '
              f'train {train_loss:.6f}, development {dev_loss:.6f}')
        if dev_loss < best_dev_loss:
            best_dev_loss = dev_loss
            best_epoch = epoch
            save_training_state(best_path, epoch)
            print('  Best weights saved to Drive.')
        save_training_state(last_path, epoch)

# For scoring and plots, always use the best validation checkpoint.
if not best_path.is_file():
    raise FileNotFoundError('Missing best student weights: ' + str(best_path))
checkpoint = torch.load(best_path, map_location=teacher.device, weights_only=True)
if checkpoint.get('run_config') != run_config:
    raise RuntimeError('The best checkpoint belongs to a different experiment.')
student_model.load_state_dict(checkpoint['model_state_dict'])
student_model.eval()
print(f'Loaded best student weights (epoch {checkpoint["epoch"]}, '
      f'dev loss {checkpoint["best_dev_loss"]:.6f}).')
print('File:', best_path)
```

## Stage 7 — Teacher–student anomaly heatmap

The frozen teacher and trained student produce a cosine-discrepancy map. This is a prototype detection score; testing across real defects is a later stage.

```python
import math
import torch.nn.functional as F


@torch.no_grad()
def anomaly_score(image):
    """
    Return:
      1. A 224 x 224 teacher-student discrepancy map
      2. An image score: mean of the highest-scoring 1% of pixels
    """

    # Teacher preprocesses the image and extracts its features
    teacher_pixels, teacher_features = teacher.extract(
        image.convert("RGB")
    )

    # Student receives the same preprocessed image
    student_features = student_model(teacher_pixels)

    if teacher_features.shape != student_features.shape:
        raise ValueError("Teacher and student feature shapes do not match.")

    # Compare the 384-element feature vectors at each grid position.
    # Shape: [1, 14, 14]
    patch_map = 1 - F.cosine_similarity(
        teacher_features,
        student_features,
        dim=1
    )

    # Resize to the processed image size: [1, 1, 224, 224]
    image_map = F.interpolate(
        patch_map.unsqueeze(1),
        size=teacher_pixels.shape[-2:],
        mode="bilinear",
        align_corners=False
    )

    # Average the highest-scoring 1% of pixels
    flattened = image_map.flatten()
    k = max(1, math.ceil(0.01 * flattened.numel()))

    score = torch.topk(flattened, k).values.mean().item()

    return image_map[0, 0].cpu(), score
```

### Calibrate a normal-image threshold for each category

Thresholds are estimated only from held-out **normal development images**, never from real defective test images. Different categories can have different normal discrepancy scores.

```python
import json

thresholds_path = checkpoint_dir / f'{checkpoint_prefix}_{run_id}_thresholds.json'
threshold_version = 'cosine_top1percent_normal95_v1'
# A later training run can replace the best checkpoint without changing its
# experiment ID. Recompute thresholds whenever the checkpoint bytes change.
checkpoint_sha256 = hashlib.sha256(best_path.read_bytes()).hexdigest()

if thresholds_path.is_file() and not FORCE_RETRAIN:
    saved_thresholds = json.loads(thresholds_path.read_text())
else:
    saved_thresholds = {}

if (saved_thresholds.get('run_id') == run_id
        and saved_thresholds.get('version') == threshold_version
        and saved_thresholds.get('checkpoint_sha256') == checkpoint_sha256
        and set(saved_thresholds.get('thresholds', {})) == set(ALL_CATEGORIES)):
    category_thresholds = saved_thresholds['thresholds']
    print('Loaded existing normal-development thresholds from Drive.')
else:
    category_thresholds = {}
    for category in ALL_CATEGORIES:
        dev_scores = []
        for path in category_dev_paths[category]:
            with Image.open(path) as image:
                _, score = anomaly_score(image)
            dev_scores.append(score)
        category_thresholds[category] = torch.quantile(
            torch.tensor(dev_scores, dtype=torch.float32), 0.95
        ).item()
        print(f'{category:13s}: {len(dev_scores):3d} normal dev images | '
              f'threshold {category_thresholds[category]:.6f}')
    thresholds_path.write_text(json.dumps({
        'run_id': run_id,
        'version': threshold_version,
        'checkpoint_sha256': checkpoint_sha256,
        'thresholds': category_thresholds,
    }, indent=2))
    print('Saved threshold estimates:', thresholds_path)

# Keep the existing bottle visualization compatible.
threshold = category_thresholds['bottle']
print('Bottle 95th-percentile normal-development threshold:', threshold)
```
